```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries Imported Successfully!")
```

```python
df = pd.read_csv("CodeAlpha_Web_Scraping.csv")

print("Dataset Shape:", df.shape)

df.head()
```

```python
plt.figure(figsize=(9,5))

sns.countplot(
    data=df,
    x="Rating",
    order=["One", "Two", "Three", "Four", "Five"]
)

plt.title("Book Rating Distribution")
plt.xlabel("Book Rating")
plt.ylabel("Number of Books")

plt.savefig("Rating_Distribution.png")

plt.show()
```

```python
plt.figure(figsize=(9,5))

plt.hist(df["Price"], bins=20)

plt.title("Book Price Distribution")
plt.xlabel("Book Price (£)")
plt.ylabel("Number of Books")

plt.savefig("Price_Distribution.png")

plt.show()
```

```python
plt.figure(figsize=(9,5))

sns.barplot(
    data=df,
    x="Rating",
    y="Price",
    order=["One", "Two", "Three", "Four", "Five"]
)

plt.title("Average Book Price by Rating")
plt.xlabel("Book Rating")
plt.ylabel("Average Price (£)")

plt.savefig("Average_Price_by_Rating.png")

plt.show()
```

```python
plt.figure(figsize=(9,5))

sns.boxplot(x=df["Price"])

plt.title("Book Price Outlier Analysis")
plt.xlabel("Book Price (£)")

plt.savefig("Price_Outliers.png")

plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(15,10))

# Rating Distribution
sns.countplot(
    data=df,
    x="Rating",
    order=["One", "Two", "Three", "Four", "Five"],
    ax=axes[0,0]
)
axes[0,0].set_title("Rating Distribution")

# Price Distribution
axes[0,1].hist(df["Price"], bins=20)
axes[0,1].set_title("Price Distribution")

# Average Price by Rating
sns.barplot(
    data=df,
    x="Rating",
    y="Price",
    order=["One", "Two", "Three", "Four", "Five"],
    ax=axes[1,0]
)
axes[1,0].set_title("Average Price by Rating")

# Price Outliers
sns.boxplot(x=df["Price"], ax=axes[1,1])
axes[1,1].set_title("Price Outlier Analysis")

plt.suptitle("BOOK DATA ANALYSIS DASHBOARD", fontsize=18)

plt.tight_layout()

plt.savefig("CodeAlpha_Dashboard.png", dpi=300)

plt.show()
```
